WsXM cant open .png so here we open it with PIL and convert to an XYZ file

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from PIL import Image
```

```python
img = Image.open('fig1a.png').convert('L') # Grayscale
img = np.array(img)
```

```python
# Gracefully written by ChatGPT4
# Get the indices and values
x, y = np.indices(img.shape)
z = img.flatten()

# Combine x, y, z into a single array
xyz = np.column_stack((x.flatten(), y.flatten(), z))

# Save to a text file
np.savetxt('fig1a.xyz', xyz, fmt='%d %d %d')
```
