```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
fdir = "/home/cmp/Documents/Github/QPI-Scattering/CAnalytics/output/single0a/onedef-ldos.tsv"
```

```python
df = pd.read_csv(fdir, sep="\t", header=None, dtype=np.double, skiprows=3)
```

```python
xlin = np.linspace(-10, 10, 1001)
ylin = np.linspace(-10, 10, 1001)
X, Y = np.meshgrid(xlin, ylin)
```

```python
# We create a mask because our functions blow up at r -> zero
maskr = 0.5
mask1 = ((X) ** 2 + Y ** 2) <= maskr**2

mask = mask1

Z_masked = df.values * (X ** 2 + Y ** 2)
Z_masked[mask] = 1e-9

fig, ax = plt.subplots(figsize=(10, 10))

#pc = ax.pcolormesh(xlin, ylin, ZR2, cmap="winter", vmax=15, vmin=-15)
#pc = ax.pcolormesh(xlin, ylin, Z_masked, cmap="winter", vmax=1.5)
pc = ax.pcolormesh(xlin, ylin, np.square(Z_masked))
fig.colorbar(pc)

ax.set_aspect('equal')
ax.set_xlabel('x (nm)')
ax.set_ylabel('y (nm)')
```
