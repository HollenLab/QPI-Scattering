```python
import glob
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

%matplotlib inline
```

```python
csv_files = glob.glob("./output/schemes/0d2eV/7x7/sep3/*.csv")

data = np.zeros((700, 700))

for file in csv_files:
    df = pd.read_csv(file, header=0, index_col=0)
    array = df.to_numpy()

    data += array

print("done :)")
```

```python
%matplotlib inline
```

```python
xlin = np.linspace(-7, 7, 700)
ylin = np.linspace(-7, 7, 700)

xx, yy = np.meshgrid(xlin, ylin)

fig, ax = plt.subplots(figsize=(8, 8))

ax.set_xlim(-6, 6)
ax.set_ylim(-6, 6)

pc = ax.pcolormesh(xx, yy, data)
cbar = fig.colorbar(pc)
ax.set_aspect('equal')
```
