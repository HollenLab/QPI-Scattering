```python
import glob
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
csv_files = glob.glob("./output/schemes/0d2eV/5x5/*.csv")

data = np.zeros((500, 500))

for file in csv_files:
    df = pd.read_csv(file, header=0, index_col=0)
    array = df.to_numpy()

    data += array

print("done :)")
```

```python
xlin = np.linspace(-3, 3, 300)
ylin = np.linspace(-3, 3, 300)

xx, yy = np.meshgrid(xlin, ylin)

fig, ax = plt.subplots(figsize=(8, 8))

ax.pcolormesh(xx, yy, np.log(data ** 2))
ax.set_aspect('equal')
```
